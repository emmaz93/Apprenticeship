```python
%%writefile src/tableau_refresh.py
import os
import tableauserverclient as TSC

def trigger_refresh(workbook_id):
  """Triggers an extract refresh for a specific Tableau workbook"""
  # These will be pulled from GitHub Secrets later
  token_name = os.getenv('TABLEAU_TOKEN_NAME')
  token_value = os.getenv('TABLEAU_TOKEN_VALUE')
  site_id = os.getenv('TABLEAU_SITE_ID')
  server_url = 'https://prod-useast-b.online.tableau.com'

  tableau_auth = TSC.PersonalAccessTokenAuth(token_name, token_value, site_id)
  server = TSC.Server(server_url, use_server_version=True)

  try:
        with server.auth.sign_in(tableau_auth):
            workbook = server.workbooks.get_by_id(workbook_id)
            server.workbooks.refresh(workbook)
            return True
  except Exception as e:
        print(f"Failed to refresh: {e}")
        return False
```

```python
%%writefile tests/test_main.py
import pytest
from tableau_refresh import trigger_refresh
import os

def test_refresh_call_logic(mocker):
    # Mock the Tableau Server connection
    mock_server = mocker.patch('tableauserverclient.Server')
    instance = mock_server.return_value

    # Mock environment variables
    mocker.patch.dict(os.environ, {
        'TABLEAU_TOKEN_NAME': 'mock_token_name',
        'TABLEAU_TOKEN_VALUE': 'mock_token_value',
        'TABLEAU_SITE_ID': 'mock_site_id'
    })

    # Mock context manager and workbook
    instance.auth.sign_in.return_value.__enter__.return_value = None
    mock_wb = mocker.MagicMock()
    instance.workbooks.get_by_id.return_value = mock_wb

    # Test the function
    success = trigger_refresh("mock-id-123")

    assert success is True
    instance.workbooks.refresh.assert_called_once_with(mock_wb)
```
